## Ejercicio 3

>
> ```ejer_3_2 = ["Que", "tal"]```

````python
ejer_3_1=["Hola","amigo"]
ejer_3_2=["Que","tal"]

ejer_3_3=[]

for i in ejer_3_1:
    for j in ejer_3_2:
        ejer_3_3.append(f"{i} {j}")
        # ejer_3_3.append(i+j)

print(ejer_3_3)
````

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

Consejo: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5=[3,20,3,47,19,3,29,45,67,78,90,3,3]
ejer_5_1=ejer_5.copy()
print(ejer_5)

for i in ejer_5:
    if i==3:
        ejer_5_1.remove(3)

print(ejer_5_1)
````

## Ejercicio 12
Convierte la siguiente tupla en un diccionario
> ```ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]
print("Tupla a diccionario:",dict(ejer_12), ",esto no funciona bien porque se repiten las x y las y")

dic12 = {}

for clave, valor in ejer_12:
    if clave not in dic12:
        dic12[clave] = []
    dic12[clave].append(valor)

print("Diccionario final:",dic12)
````
