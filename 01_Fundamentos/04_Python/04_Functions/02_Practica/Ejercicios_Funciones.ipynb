## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def semana(num):
    if num == 1:
        return "lunes"
    elif num == 2:
        return "martes"
    elif num == 3:
        return "miercoles"
    elif num == 4:
        return "jueves"
    elif num == 5:
        return "viernes"
    elif num == 6:
        return "sabado"
    elif num == 7:
        return "domingo"
    else:
        return "Numero de la semana no valido"


semana(4)
````

## Ejercicio 3
Escibe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero se  mayor que el segundo, o que el segundo sea mayor que el primero

````python

def comparacion(n1, n2):
    if n1>n2:
        return print(n1, "es mayor que", n2)
    elif n1 == n2:
        return print(n1, "es igual que", n2)
    elif n1<n2:
        return print(n1, "es menor que", n2)
    else:
        return print("Datos no válidos")

comparacion(4,3)
comparacion(3,9)
comparacion(6,6)
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string.

````python
def contar_letras(texto):
    conteo = {}
    for letra in texto:
        if letra in conteo:
            conteo[letra] += 1
        else:
            conteo[letra] = 1
    return conteo


contar_letras("supercalifragisiticoespialidoso")
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
def fibonacci(n):
    if n <= 1:
        return n
    
    return fibonacci(n - 1) + fibonacci(n - 2)
        

fibonacci(9)
````

````python
def cuadrado(base):
    area= base**2
    print("El área de un cuadrado de base", base,"es: ", area)
    return area

def triangulo(base_triangulo, altura_triangulo):
    area= (base_triangulo*altura_triangulo)/2
    print("El área de un triangulo de base", base_triangulo, "y altura", altura_triangulo,"es: ", area)
    return area

import math
def circulo(radio):
    area= (radio**2)*math.pi
    print("El área de un circulo de radio", radio,"es: ", area)
    return area
````
